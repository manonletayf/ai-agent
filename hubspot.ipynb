```python
import hubspot
from pprint import pprint
from hubspot.crm.lists import ApiException
from dotenv import load_dotenv
import os

load_dotenv()
hubspot_access_token = os.getenv("HUBSPOT_ACCESS_TOKEN")
client = hubspot.Client.create(access_token=hubspot_access_token)
```

```python
try:
    api_response = client.crm.lists.lists_api.get_by_id(list_id=20329, include_filters=True)
    print(api_response)
except ApiException as e:
    print("Exception when calling lists_api->get_by_id: %s\n" % e)
```
